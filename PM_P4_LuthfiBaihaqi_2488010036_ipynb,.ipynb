```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# ==========================================
# LANGKAH 1 — Menangani nilai hilang
# ==========================================

df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())


# ==========================================
# LANGKAH 2 — Memisahkan fitur (X) dan label (y)
# ==========================================

X = df.drop(columns=['membeli'])
y = df['membeli']


# ==========================================
# LANGKAH 3 — Encoding kategorikal
# ==========================================

# Encoding pendidikan (ordinal)
X['pendidikan'] = X['pendidikan'].map({
    'SMA': 0,
    'S1': 1,
    'S2': 2
})

# One-Hot Encoding kota jika kolom kota tersedia
if 'kota' in X.columns:
    X = pd.get_dummies(
        X,
        columns=['kota'],
        dtype=int
    )

# Encoding label
y = y.map({
    'Tidak': 0,
    'Ya': 1
})


# ==========================================
# LANGKAH 4 — Membagi data
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)


# ==========================================
# LANGKAH 5 — Penskalahan
# ==========================================

num = ['usia', 'pendapatan', 'pendidikan']

sc = StandardScaler()

X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])


# ==========================================
# LANGKAH 6 — Verifikasi
# ==========================================

print("===== VERIFIKASI =====")

print("\n1. Missing value:")
print(X.isnull().sum())

print("\n2. Data X:")
print(X)

print("\n3. Data y:")
print(y)

print("\n4. Data training:")
print(X_train)

print("\n5. Data testing:")
print(X_test)

print("\n6. Statistik data training:")
print(X_train[num].describe())
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
data = {
    'usia': [20, 25, 30, 35, 40],
    'pendapatan': [3000000, 4000000, 5000000, 6000000, 7000000],
    'pendidikan': ['SMA', 'S1', 'S1', 'S2', 'S2'],
    'membeli': ['Tidak', 'Ya', 'Tidak', 'Ya', 'Ya']
}

df = pd.DataFrame(data)

df['pendidikan'] = df['pendidikan'].map({
    'SMA': 0,
    'S1': 1,
    'S2': 2
})

df['membeli'] = df['membeli'].map({
    'Tidak': 0,
    'Ya': 1
})

X = df.drop(columns=['membeli'])
y = df['membeli']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

sc = MinMaxScaler()

num = ['usia', 'pendapatan', 'pendidikan']

X_train[num] = sc.fit_transform(X_train[num])

X_test[num] = sc.transform(X_test[num])

print("Data training setelah MinMaxScaler:")
print(X_train[num])

print("\nDeskripsi data training:")
print(X_train[num].describe())
```

## Latihan 4

fit_transform() digunakan pada data latih karena scaler perlu mempelajari parameter seperti nilai minimum dan maksimum dari data latih sekaligus melakukan transformasi. Sementara itu, data uji hanya menggunakan transform() agar menggunakan parameter yang telah dipelajari dari data latih. Hal ini dilakukan untuk mencegah data leakage, sehingga data uji tetap benar-benar independen dari proses pembelajaran.

## Refleksi

1. Mengapa urutan split dulu, baru scaling penting?
Urutan split dulu kemudian scaling penting untuk mencegah terjadinya data leakage. Data training digunakan untuk menentukan parameter scaling, sedangkan data testing hanya menggunakan parameter yang sudah diperoleh dari data training. Dengan begitu, data testing tetap benar-benar menjadi data yang belum diketahui oleh model.
2. Kapan label encoding dan kapan one-hot encoding?
Label encoding digunakan ketika data kategorikal memiliki urutan atau tingkatan, seperti tingkat pendidikan SMA, S1, dan S2. Sedangkan one-hot encoding digunakan ketika data kategorikal tidak memiliki urutan, seperti nama kota, jenis kelamin, atau warna.
3. Apa perbedaan normalisasi dan standardisasi?
Normalisasi adalah proses mengubah nilai data ke dalam rentang tertentu, biasanya 0 sampai 1, sedangkan standardisasi mengubah data berdasarkan nilai rata-rata dan standar deviasi sehingga rata-ratanya mendekati 0 dan standar deviasinya mendekati 1.
